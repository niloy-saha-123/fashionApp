```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Load the MobileNetV2 model pretrained on ImageNet
model = MobileNetV2(weights="imagenet")
```

```python
# Load the image
img_path = "../datasets/sneaker.jpg"

img = image.load_img(img_path, target_size=(224, 224))  # Resize to 224x224

# Convert image to numpy array
img_array = image.img_to_array(img)

# Expand dimensions (MobileNet expects batch format)
img_array = np.expand_dims(img_array, axis=0)

# Preprocess the image for MobileNet
img_array = preprocess_input(img_array)

# Display the processed image
plt.imshow(img)
plt.axis("off")
plt.title("Input Image")
plt.show()
```

```python
# Predict class label
predictions = model.predict(img_array)

# Decode top 3 predictions
decoded_predictions = decode_predictions(predictions, top=3)[0]

# Print results
for i, (imagenet_id, label, score) in enumerate(decoded_predictions):
    print(f"{i+1}: {label} ({score:.2f})")
```
